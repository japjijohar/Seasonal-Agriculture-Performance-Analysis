```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("seasonal_agriculture_performance_dataset.csv")
print("dataset loaded successfully!!")

df.head()
```

```python
df.shape
df.info()
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df.fillna(df.select_dtypes(include=np.number).mean(), inplace=True)
```

```python
# for categorical values:
for col in df.select_dtypes(include='object').columns:
    df[col]= df[col].fillna(df[col].mode()[0])
print("missing values handled!!")
```

```python
df.isnull().sum()
```

```python
print("duplicate records are:", df.duplicated().sum())
```

```python
#checking for unique values 
for col in df.columns:
    print("\n", col)
    print(df[col].unique())
```

```python
#univariate analysis
plt.figure(figsize=(8,5))
plt.hist(df['Yield_Tonnes'], bins=20)
plt.xlabel('yield')
plt.ylabel('frequency')
plt.title('crop yield distribution')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
#univariate analysis
plt.figure(figsize=(8,5))
plt.hist(df['Yield_Tonnes_Ha'], bins=20)
plt.xlabel('yield')
plt.ylabel('frequency')
plt.title('crop yield distribution')
plt.show()
```

```python
#univariate analysis: box plot
plt.figure(figsize=(8,5))
plt.boxplot(x=df['Yield_Tonnes_Ha'])
plt.xlabel('yield')
plt.title('crop yield distribution box plot')
plt.show()
```

```python
#univariate analysis: season distribution 
plt.figure(figsize=(8,5))
df['Season'].value_counts().plot(kind='bar')
plt.xlabel('season')
plt.ylabel('number of farms')
plt.title('number of farms by season')
plt.show()
```

```python
#bivariate analysis:
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    data= df
)
plt.xlabel('rainfall')
plt.ylabel('yields')
plt.title('rainfall vs crop yield')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Avg_Temperature_C',
    y='Yield_Tonnes_Ha',
    data= df
)
plt.xlabel('temperature')
plt.ylabel('yields')
plt.title('temperature vs crop yield')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='Water_Used_m3',
    y='Yield_Tonnes_Ha',
    data= df
)
plt.xlabel('water used')
plt.ylabel('yields')
plt.title('water used vs crop yield')
plt.show()
```

```python
#average yield by season
season_yield= df.groupby('Season')['Yield_Tonnes_Ha'].mean()
print(season_yield)
```

```python
#graph for seasonal yields
plt.figure(figsize=(8,5))
season_yield.plot(kind='bar')
plt.xlabel('season')
plt.ylabel('average yield')
plt.title('crop yield by season')
plt.show()
```

```python
#finding the best season:
best_season= season_yield.idxmax()
print("best season:", best_season)
print("average yields:", season_yield.max())
```

```python
# crop analysis:
crop_yield= df.groupby('Crop')['Yield_Tonnes_Ha'].mean()
print(crop_yield)
```

```python
#graph
plt.figure(figsize=(10,5))
crop_yield.plot(kind='bar')
plt.xlabel('crop')
plt.ylabel('average yield')
plt.title('avg yield by crop')
plt.xticks(rotation=45)
plt.show()
```

```python
#best crops:
print("best crops:", crop_yield.idxmax())
print("average yield:", crop_yield.max())
```

```python
crop_season= df.groupby(
    ['Crop','Season']
)['Yield_Tonnes_Ha'].mean()
print(crop_season)
```

```python
best_combination= crop_season.idxmax()
print("best crop season combinations:", best_combination)
print("average yield:", crop_season.max())
```

```python
# correlation analysis:
numeric_df= df.select_dtypes(include=np.number)
correlation= numeric_df.corr()
plt.figure(figsize=(14,10))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)
plt.title('correlation heatmap')
plt.show()
```
